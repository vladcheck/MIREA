```python
# 1.2
import sympy as sp

augmented_matrix = sp.Matrix([
    [5, -1, 5,  9, -2, -3],
    [3,  0, 6,  9, -1,  4],
    [2,  0, 4,  6, -1,  0]
])

print("Исходная расширенная матрица:")
sp.pprint(augmented_matrix)

rref_matrix, pivot_columns = augmented_matrix.rref()

print("а) rref(A|b) =")
sp.pprint(rref_matrix)

# Разделение расширенной матрицы на матрицу коэффициентов A и вектор свободных членов b
A = augmented_matrix[:, :-1]
b = augmented_matrix[:, -1]

x1, x2, x3, x4, x5 = sp.symbols('x1 x2 x3 x4 x5')

solution = sp.linsolve((A, b), (x1, x2, x3, x4, x5))

print("б) Общее решение системы:")
sp.pprint(tuple(solution)[0]) # type: ignore
```

```python
# 1.3
import sympy as sp

x, y, z, u = sp.symbols('x y z u')

eq1 = sp.Eq(32*x + 97*y + 90*z + 48*u, 47)
eq2 = sp.Eq(35*x + 84*y + 91*z + 54*u, 40)
eq3 = sp.Eq(17*x + 18*y + 21*z + 13*u, 37)

solution = sp.linsolve([eq1, eq2, eq3], (x, y, z, u))

sol_tuple = list(solution)[0] # type: ignore
y_expr = sol_tuple[1]

print("Зависимость y от x:")
sp.pprint(y_expr)

poly_y = sp.Poly(y_expr, x)
coeffs = poly_y.all_coeffs()
print(coeffs)

b_val = coeffs[0]
a_val = coeffs[1]

print(f"1) a = {a_val}")
print(f"2) b = {b_val}")
```

```python
# 1.6
import numpy as np
from itertools import product

def max_det_order_6():
    n = 6
    max_determinant = 0
    best_matrix = None
    
    first_row = np.ones(n, dtype=np.int8)
    all_rows = list(product([-1, 0, 1], repeat=n))
    valid_rows = [np.array(r, dtype=np.int8) for r in all_rows if any(v != 0 for v in r)]
    
    checked = 0
    
    def search(current_matrix, depth):
        nonlocal max_determinant, best_matrix, checked
        
        if depth == n:
            det = abs(round(np.linalg.det(current_matrix)))
            if det > max_determinant:
                max_determinant = det
                best_matrix = current_matrix.copy()
            return
        
        for row in valid_rows:
            current_matrix[depth] = row
            
            if depth >= 1:
                rank = np.linalg.matrix_rank(current_matrix[:depth+1])
                if rank < depth + 1:
                    continue
            
            search(current_matrix, depth + 1)
    
    matrix = np.zeros((n, n), dtype=np.int8)
    matrix[0] = first_row
    
    print("Запуск поиска...")
    search(matrix, 1)
    
    return max_determinant, best_matrix

max_det, optimal_matrix = max_det_order_6()
print(f"Максимальный определитель: {max_det}")
print("Пример оптимальной матрицы:")
print(optimal_matrix)
```

```python
# 1.9
import numpy as np

I = np.array([[1,0],[0,1]])
gamma = 0.057
A = np.array([[7,14],[9,1]]) * gamma

def Smn(m:int,n:int):
  sum = np.zeros([2,2])
  for i in range(m,n+1):
    sum += A**i
  return sum

def Bm(m:int):
  return np.linalg.matmul(np.linalg.inv(I-A),A**m)

S = Smn(4,70)
print(np.max(S))

B = Bm(4)
print(np.max(B))
```
