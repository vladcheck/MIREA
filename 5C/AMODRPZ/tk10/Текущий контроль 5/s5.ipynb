```python
# создание класса perceptron
class Perceptron(torch.nn.Module):
    def __init__(self, input_dim=784, num_layers=0,
            hidden_dim=64, output_dim=10, p=0.0, device='cpu'): 
        super(Perceptron, self).__init__()
        self.layers = torch.nn.Sequential()
        prev_size = input_dim 
        for i in range(num_layers):
            self.layers.add_module('layer{}'.format(i), torch.nn.Linear(prev_size, hidden_dim)) 
            self.layers.add_module('relu{}'.format(i), torch.nn.ReLU())
            self.layers.add_module('dropout{}'.format(i), torch.nn.Dropout(p=p)) 
            prev_size = hidden_dim
        self.layers.add_module('classifier',
            torch.nn.Linear(prev_size, output_dim))
        self.to(device)

    def forward(self, input):
        return self.layers(input)
```

```python
# это объект, обозначающий устройство, на котором размещен или будет размещен torch.Tensor \
# Torch.device содержит тип устройства («процессор» или « cuda»)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu") 
device
```

```python
# Создание модели без внутренних слоев 
model = Perceptron(device=device) 
model
```

```python
# процедура терстирования модели 
def testing(model, dataset):
    generator = torch.utils.data.DataLoader(dataset, batch_size=64)
    pred = []
    real = []
    for x, y in tqdm(generator, leave=False):
        x = x.view([-1, 784]).to(device) 
        y = y.to(device)
        pred.extend(torch.argmax(model(x), dim=-1).cpu().numpy().tolist()) 
        real.extend(y.cpu().numpy().tolist())
    
    return np.mean(np.array(real) == np.array(pred)), classification_report(real, pred)
```

```python
# процедура тренировки модели
def trainer(model, dataset, loss_function, optimizer, epochs): 
    for epoch in tqdm(range(epochs), leave=False):
        generator = torch.utils.data.DataLoader(dataset, batch_size=64,
            shuffle=True)

        for x, y in tqdm(generator, leave=False):
            optimizer.zero_grad() 
            x = x.view([-1, 784]).to(device) 
            y = y.to(device)
            output = model(x)
            loss = loss_function(output, y)
            loss.backward()
            optimizer.step()
```

```python
model = Perceptron(num_layers=0, device=device)
```

```python
# Результат, получаемый от необученной модели.
_ = model.eval()
acc, report = testing(model, MNIST_test) 
print(report)
```

```python
# Обучение модели
_ = model.train()
trainer(model=model,
    dataset=MNIST_train,
    loss_function=torch.nn.CrossEntropyLoss(),
    optimizer=torch.optim.Adam(model.parameters(), lr=0.001), 
    epochs=4)
```

```python
# Результат обучения классификатора
_ = model.eval()
acc, report = testing(model, MNIST_test) 
print(report)
```

Задание: С помощью нейронной сети классифицировать вина по качеству. Датасет взять из третьей работы. Разбить датасет на два класса по качеству вина. Качество вина до пяти включительно определить как первый класс, свыше пяти - как второй.
Представить основные показатели метода. Сравнить результат классификации с результатом из третьей работы.

```python
# Задание: с помощью нейронной сети классифицировать вина по качеству.
# Датасет взят из третьей работы: winequality-red.csv (лежит в каталоге data/).
import pandas as pd
from torch.utils.data import TensorDataset
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, f1_score
from sklearn.linear_model import LogisticRegression

torch.manual_seed(42)  # фиксация случайных зерен для воспроизводимости результата
np.random.seed(42)

df = pd.read_csv('./data/winequality-red.csv')  # открытие датасета
print("размер датасета:", df.shape)
print("качество вина:", df["quality"].value_counts().sort_index().to_dict())

# Разбиваем датасет на два класса по качеству вина: до пяти включительно —
# первый класс (0), свыше пяти — второй класс (1).
y = (df["quality"] > 5).astype(int).values
print("классы 0/1:", np.bincount(y))
X = df.drop(columns=["quality"]).values  # 11 физико-химических признаков

# разбиение датасета на тестовый и обучающий (80/20, с сохранением долей классов)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# стандартизация признаков: средние и дисперсии считаются только по обучающей выборке
scaler = StandardScaler().fit(Xtr)
Xtr, Xte = scaler.transform(Xtr), scaler.transform(Xte)
print("обучающая выборка:", Xtr.shape, "тестовая выборка:", Xte.shape)

# датасеты вин для обучения и тестирования
wine_train = TensorDataset(torch.tensor(Xtr, dtype=torch.float32),
                           torch.tensor(ytr, dtype=torch.long))
wine_test = TensorDataset(torch.tensor(Xte, dtype=torch.float32),
                          torch.tensor(yte, dtype=torch.long))
```

```python
# процедура тестирования модели на винах (как testing, но признаки уже плоские:
# view(x.size(0), -1) вместо view([-1, 784]) для картинок MNIST)
def testing_wine(model, dataset):
    generator = torch.utils.data.DataLoader(dataset, batch_size=64)
    pred = []
    real = []
    for x, y in tqdm(generator, leave=False):
        x = x.view(x.size(0), -1).to(device)
        y = y.to(device)
        pred.extend(torch.argmax(model(x), dim=-1).cpu().numpy().tolist())
        real.extend(y.cpu().numpy().tolist())

    return np.mean(np.array(real) == np.array(pred)), classification_report(real, pred)


# процедура тренировки модели на винах (как trainer, но дополнительно возвращает
# историю loss по эпохам для построения кривой обучения)
def trainer_wine(model, dataset, loss_function, optimizer, epochs):
    history = []  # средний loss каждой эпохи
    for epoch in tqdm(range(epochs), leave=False):
        generator = torch.utils.data.DataLoader(dataset, batch_size=64,
            shuffle=True)
        total = 0.0
        for x, y in tqdm(generator, leave=False):
            optimizer.zero_grad()
            x = x.view(x.size(0), -1).to(device)
            y = y.to(device)
            output = model(x)
            loss = loss_function(output, y)
            loss.backward()
            optimizer.step()
            total += loss.item() * len(x)
        history.append(total / len(dataset))
        print(f"эпоха {epoch + 1}/{epochs}, loss: {history[-1]:.4f}")
    return history
```

```python
# Создание модели для вин: 11 признаков -> скрытый слой 64 -> 2 класса
wine_model = Perceptron(input_dim=11, num_layers=1, hidden_dim=64, output_dim=2, device=device)
print(wine_model)

# Обучение модели
_ = wine_model.train()
loss_hist = trainer_wine(model=wine_model,
    dataset=wine_train,
    loss_function=torch.nn.CrossEntropyLoss(),
    optimizer=torch.optim.Adam(wine_model.parameters(), lr=0.001),
    epochs=25)

# Кривая обучения: номер эпохи -> средний loss, падение означает, что сеть учится
plt.figure()
plt.plot(range(1, len(loss_hist) + 1), loss_hist, marker="o")
plt.xlabel("Эпоха")
plt.ylabel("Средний loss (CrossEntropy)")
plt.title("Кривая обучения персептрона на винах")
plt.grid(True)
plt.show()
print(f"loss: {loss_hist[0]:.4f} -> {loss_hist[-1]:.4f}")
```

```python
# Результат обучения классификатора
_ = wine_model.eval()
wine_acc, wine_report = testing_wine(wine_model, wine_test)
print(f"accuracy: {wine_acc:.4f}")
print(wine_report)

# Матрица ошибок: строки — истинный класс, столбцы — предсказанный класс
wine_pred = []
with torch.no_grad():
    for x, _ in torch.utils.data.DataLoader(wine_test, batch_size=64):
        wine_pred.extend(wine_model(x.view(x.size(0), -1).to(device)).argmax(dim=1).cpu().tolist())
wine_pred = np.array(wine_pred)
cm = confusion_matrix(yte, wine_pred)
print("матрица ошибок:\n", cm)
disp = ConfusionMatrixDisplay(cm, display_labels=["низкое (до 5)", "высокое (свыше 5)"])
disp.plot()
plt.title("Матрица ошибок персептрона (вина)")
plt.show()

# Сравнение с результатом из третьей работы: логистическая регрессия на том же разбиении
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(Xtr, ytr)
lr_pred = lr.predict(Xte)
lr_acc = np.mean(yte == lr_pred)
print("=== Логистическая регрессия (третья работа, тест) ===")
print(f"accuracy: {lr_acc:.4f}")
print(classification_report(yte, lr_pred, target_names=["низкое (до 5)", "высокое (свыше 5)"]))
print(f"персептрон: accuracy={wine_acc:.4f}, f1={f1_score(yte, wine_pred):.4f}")
print(f"логистическая регрессия: accuracy={lr_acc:.4f}, f1={f1_score(yte, lr_pred):.4f}")
```
